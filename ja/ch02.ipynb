## 10. 行数のカウント
ファイルの行数をカウントせよ。確認にはwcコマンドを用いよ。

```python
def count(file):
   with open(file) as f:
      lines = f.readlines()
      return len(lines)
```

```python
print(count('/content/drive/MyDrive/popular-names.txt'))
```

```python
!wc /content/drive/MyDrive/popular-names.txt
```

## 11. 先頭からN行を出力
ファイルの先頭N行だけを表示せよ。例えば、N=10として先頭10行を表示せよ。確認にはheadコマンドを用いよ。

```python
def head(file, n):
  output = []
  with open(file) as f:
    for i, line in enumerate(f):
      if i < n:
        output.append(line)
  return ''.join(output)
```

```python
print(head('/content/drive/MyDrive/popular-names.txt', 10))
```

```python
!head -n 10 /content/drive/MyDrive/popular-names.txt
```

## 12. 末尾のN行を出力
ファイルの末尾N行だけを表示せよ。例えば、N=10として末尾10行を表示せよ。確認にはtailコマンドを用いよ。

```python
def tail(file, n):
  output = []
  with open(file) as f:
    start = count(file) - n
    lines = f.readlines()
    for i, line in enumerate(lines):
      if start-1 < i < count(file):
        output.append(line)
  return ''.join(output)
```

```python
print(tail('/content/drive/MyDrive/popular-names.txt', 10))
```

```python
!tail -n 10 /content/drive/MyDrive/popular-names.txt
```

## 13. タブをスペースに置換
ファイルの先頭10行に対して、タブ1文字につきスペース1文字に置換して出力せよ。確認にはsedコマンド、trコマンド、もしくはexpandコマンドなどを用いよ。

```python
lines = head('/content/drive/MyDrive/popular-names.txt', 10)

for s in lines.replace('\t', ' '):
  print(s, end='')
```

```python
!head -n 10 /content/drive/MyDrive/popular-names.txt | sed 's/\t/ /g'
```

```python
lines = head('/content/drive/MyDrive/popular-names.txt', 10)

for s in lines.split('\n'):
    print(s.split('\t')[0])
```

```python
!head -n 10 /content/drive/MyDrive/popular-names.txt |cut -f 1
```
