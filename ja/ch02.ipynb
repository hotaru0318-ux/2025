## 10. 行数のカウント
ファイルの行数をカウントせよ。確認にはwcコマンドを用いよ。

```python
def count(file):
   with open(file, 'r') as f:
      lines = f.readlines()
      return len(lines)
```

```python
print(count('/content/drive/MyDrive/popular-names.txt'))
```

```python
!wc /content/drive/MyDrive/popular-names.txt
```

## 11. 先頭からN行を出力
ファイルの先頭N行だけを表示せよ。例えば、N=10として先頭10行を表示せよ。確認にはheadコマンドを用いよ。

```python
def head(file, n):
  output = []
  with open(file, 'r') as f:
    for i, line in enumerate(f):
      if i < n:
        output.append(line)
  return ''.join(output)
```

```python
print(head('/content/drive/MyDrive/popular-names.txt', 10))
```

```python
!head -n 10 /content/drive/MyDrive/popular-names.txt
```

## 12. 末尾のN行を出力
ファイルの末尾N行だけを表示せよ。例えば、N=10として末尾10行を表示せよ。確認にはtailコマンドを用いよ。

```python
def tail(file, n):
  output = []
  with open(file, 'r') as f:
    start = count(file) - n
    lines = f.readlines()
    for i, line in enumerate(lines):
      if start-1 < i < count(file):
        output.append(line)
  return ''.join(output)
```

```python
print(tail('/content/drive/MyDrive/popular-names.txt', 10))
```

```python
!tail -n 10 /content/drive/MyDrive/popular-names.txt
```

## 15. ファイルをN分割する
ファイルを行単位でN分割し、別のファイルに格納せよ。例えば、N=10としてファイルを10分割せよ。同様の処理をsplitコマンドで実現せよ。

```python
def split(file, n, output_prefix="x"):
  with open(file, 'r') as f:
    lines = f.readlines()
  total = len(lines)
  sp = total // n
  remainder = total % n

  start_index = 0
  for i in range(n):
    end_index = start_index + sp
    if i < remainder: # Distribute remainder lines among the first 'remainder' files
      end_index += 1

    # Generate output filename (e.g., xa, xb, xc, ...) compatible with 'split' command naming
    output_filename = f"{output_prefix}{chr(ord('a') + i)}"
    with open(output_filename, 'w') as out_f:
      out_f.writelines(lines[start_index:end_index])
    start_index = end_index
```

```python
split('/content/drive/MyDrive/popular-names.txt', 10)
!ls xa*
```

## 16. ランダムに各行を並び替える

## 17. １列目の文字列の異なり
1列目の文字列の異なり（文字列の種類）を求めよ。確認にはcut, sort, uniqコマンドを用いよ。

```python
names = []
with open('/content/drive/MyDrive/popular-names.txt', 'r') as f:
    for line in f:
        if line.strip():
            names.append(line.split('\t')[0])

print(len(set(names)))
print(set(names))
```

```python
!cut -f 1 /content/drive/MyDrive/popular-names.txt | sort | uniq | wc -l
```

## 18. 各行の1列目の文字列の出現頻度を求め、出現頻度の高い順に並べる
1列目の文字列の出現頻度を求め、出現頻度と名前を出現頻度の多い順に並べて表示せよ。確認にはcut, uniq, sortコマンドを用いよ。

```python
from collections import Counter
counts = Counter(names)
print(counts)
```

```python
!cut -f 1 /content/drive/MyDrive/popular-names.txt | sort | uniq -c | sort -nr
```
